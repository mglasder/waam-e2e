```python
from pathlib import Path

import matplotlib.pyplot as plt

from e2e.data.loader import SampleLoader
DATA_DIR = Path("/Users/magnus/datasets/WAAM/TrainingDev")
loader = SampleLoader(DATA_DIR)
cross_section_samples = loader.load(which="all")
```

```python
sample = cross_section_samples[9]
```

```python
mean_area = 0
for sample in cross_section_samples:
    mean_area += sample.area
mean_area /= len(cross_section_samples)
mean_area
```

```python
image = sample.before_image
```

```python
import matplotlib.pyplot as plt

torch_idx_x = sample.torchposition.global_y_idx
torch_idx_z = sample.torchposition.global_z_idx

hw = 224 // 2
# greyscale
fig = plt.imshow(image[torch_idx_z - hw: torch_idx_z + hw, torch_idx_x - hw: torch_idx_x + hw ], cmap="gray")
```

```python
# save image
fig.figure.savefig("cross_section.png")
```
