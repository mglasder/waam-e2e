```python
import torch
import matplotlib.pyplot as plt

def gaussian_weights(init_weights, s):
    weights = torch.exp(-init_weights**2 / (2 * s**2))
    weights /= weights.sum()  # Normalize
    return weights

# Plotting Gaussian weights for different window sizes
sigmas = [0.3, 0.5, 1.0, 1.5,3, 5]
window_size = 15
x = torch.linspace(-(window_size // 2), window_size // 2, steps=window_size)

plt.figure(figsize=(10, 6))
for s in sigmas:
    y = gaussian_weights(x, s)
    plt.plot(x, y, label=f'Sigma {s}')

plt.title('Gaussian Weights for Different Sigmas')
plt.xlabel('Position')
plt.ylabel('Weight')
plt.legend()
plt.grid(True)
plt.show()
```
