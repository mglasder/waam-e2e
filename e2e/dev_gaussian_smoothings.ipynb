```python
import torch
import matplotlib.pyplot as plt

def gaussian_weights(window_size):
    sigma = window_size / 6  # A heuristic to relate window size to sigma
    lin_space = torch.linspace(-(window_size // 2), window_size // 2, steps=window_size)
    weights = torch.exp(-lin_space**2 / (2 * sigma**2))
    weights /= weights.sum()  # Normalize
    return lin_space, weights

# Plotting Gaussian weights for different window sizes
window_sizes = [3, 7, 15]

plt.figure(figsize=(10, 6))
for ws in window_sizes:
    x, y = gaussian_weights(ws)
    plt.plot(x, y, label=f'Window Size {ws}')

plt.title('Gaussian Weights for Different Window Sizes')
plt.xlabel('Position')
plt.ylabel('Weight')
plt.legend()
plt.grid(True)
plt.show()
```
