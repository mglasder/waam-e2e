```python
from nest.engineering.geometry import Point2
from e2e.data.sample import Mesh2D

torch_idx = s.torchposition.global_y_idx

xs = s.slice_aligned_before.xs
ys = s.slice_aligned_before.ys

h0 = ys[torch_idx]

xs_based = xs - xs[0]

slice_based_before = Mesh2D(points=[Point2.xy(x, y) for x, y in zip(xs_based, ys)])

# do the same for slice_aligned_after

xs = s.slice_aligned_after.xs
ys = s.slice_aligned_after.ys

xs_based = xs - xs[0]

slice_based_after = Mesh2D(points=[Point2.xy(x, y) for x, y in zip(xs_based, ys)])

s.slice_based_before = slice_based_before
s.slice_based_after = slice_based_after
```

```python
from e2e.data.sample import FootprintEdge

fp = s.footprint

left_based = Point2.xy(fp.left.x, fp.left.y - h0)
right_based = Point2.xy(fp.right.x, fp.right.y - h0)

fp_based = FootprintEdge(left=left_based, right=right_based, left_idx=fp.left_idx, right_idx=fp.right_idx)
s.footprint_based = fp_based
```
