```python
import torch
```

```python

# load results from file
mean_prediction = torch.load("mean_prediction.pt")
prediction_std = torch.load("prediction_std.pt")
inputs = torch.load("x.pt")
targets = torch.load("y.pt")
ids = torch.load("ids.pt")

```

```python
# plot mean and std for each sample in the batch
import matplotlib.pyplot as plt
import numpy as np
```

```python

def _plot_example(pred_mean, pred_std, trgt, inpt, id_, batch_loss, epoch, stage):
    fig, ax = plt.subplots()
    ax.plot(pred_mean.cpu().detach().numpy(), color="blue", label="pred_mean")
    # plot std around mean
    ax.fill_between(
        np.arange(len(pred_mean)),
        pred_mean.cpu().detach().numpy() - pred_std[0].cpu().detach().numpy(),
        pred_mean.cpu().detach().numpy() + pred_std[0].cpu().detach().numpy(),
        color="red",
        alpha=0.2,
        label="std",
    )
    ax.plot(trgt.cpu().numpy(), color="green", label="target (after)")
    ax.plot(inpt.cpu().detach().numpy(), color="black", label="input (before)")
    caption = f"{stage} sample: {id_} - epoch: {epoch} \n batch_loss: {batch_loss:.4f}"
    ax.set_title(caption)
    ax.legend()
```

```python
for b in range(16):
    pred_mean = mean_prediction[b]
    pred_std = prediction_std[b]
    trgt = targets[b]
    inpt = inputs[b]
    id_ = ids[b]
    batch_loss = 0
    epoch = 0
    stage = "train"
    _plot_example(pred_mean, pred_std, trgt, inpt, id_, batch_loss, epoch, stage)
    plt.show()
```
