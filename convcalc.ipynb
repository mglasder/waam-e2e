```python
import math
def calculatePadding(L, KSZ, S, D, deconv = False):
    '''
    :param L:       Input length (or width)
    :param KSZ:     Kernel size (or width)
    :param S:       Stride
    :param D:       Dilation Factor
    :return:        Returns padding such that output width is exactly half of input width
    '''
    if not deconv:
        return math.ceil((S * (L/2) - L + D * (KSZ - 1)-1)/2)
    else:
        print(L, S, D, KSZ)
        pad = math.ceil(((L - 1) * S + D * (KSZ - 1) + 1 - L * 2)/2)
        print("PAD", pad)
        output_size = (L - 1) * S - 2 * pad + D * (KSZ - 1) + 1
        print("OUTPUT SIZE", output_size)
        return pad
```

```python
# write a function as above, but so that the padding is such that the output size equals a desired output size
def calculatePadding2(L, KSZ, S, D, L_out, deconv = False):
    '''
    :param L:       Input length (or width)
    :param KSZ:     Kernel size (or width)
    :param S:       Stride
    :param D:       Dilation Factor
    :return:        Returns padding such that output width is exactly half of input width
    '''
    if not deconv:
        return math.ceil((S * (L_out) - L + D * (KSZ - 1)-1)/2)
    else:
        print(L, S, D, KSZ)
        pad = math.ceil(((L - 1) * S + D * (KSZ - 1) + 1 - L_out)/2)
        print("PAD", pad)
        output_size = (L - 1) * S - 2 * pad + D * (KSZ - 1) + 1
        print("KERNEL SIZE", KSZ)
        print("STRIDE", S)
        print("DILATION", D)
        print("OUTPUT SIZE", output_size)
        return pad
```
